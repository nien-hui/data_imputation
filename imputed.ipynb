```python
import sys
import pandas as pd
import numpy as np

if 'remasker-main' not in sys.path: 
    sys.path.append('remasker_main') 

from remasker_impute import ReMasker 
```

```python
combined_data = pd.read_csv("combined.csv")
```

```python
id_column = 'SK_ID_CURR'
```

```python
combined_data.dtypes.value_counts() 
```

```python
import pandas as pd

df = combined_data
df_result = df.copy()
threshold = 0.01

for col in df.columns:
    value_counts = df[col].value_counts(normalize=True)
    valid_categories = value_counts[value_counts >= threshold].index
    total_categories = len(value_counts)

    # 所有欄位都檢查二元比例過低的情況
    if total_categories == 2 and len(valid_categories) < 2:
        df_result = df_result.drop(columns=[col])
        continue

    if df[col].dtype == 'object':
        if len(valid_categories) > 0:
            dummies = pd.get_dummies(df[col], dummy_na=False)
            valid_existing = [c for c in valid_categories if c in dummies.columns]
            if valid_existing:
                dummies = dummies[valid_existing]
                dummies.columns = [f"{col}_{c}" for c in dummies.columns]
                df_result = pd.concat([df_result, dummies], axis=1)

        df_result = df_result.drop(columns=[col])

# print(df_result.head())
combined_data = df_result
```

### 尚未有check_point

```python
bool_columns = [col for col in combined_data.select_dtypes(include='bool').columns if col != id_column]
if bool_columns:
    print(f'將 bool 欄位轉為 float32: {len(bool_columns)}')
    combined_data[bool_columns] = combined_data[bool_columns].astype('float32')

numeric_columns = [col for col in combined_data.select_dtypes(include='number').columns if col != id_column]
numeric_data = combined_data[numeric_columns].astype('float32')

argv_backup = sys.argv.copy()
sys.argv = [sys.argv[0], '--max_epochs', '100', '--batch_size', '256']
try:
    remasker = ReMasker()
finally:
    sys.argv = argv_backup

print('啟動 ReMasker 進行填補，可能需要一些時間...') 
imputed_array = remasker.fit_transform(numeric_data) 
```

### 已有check_point

```python
bool_columns = [col for col in combined_data.select_dtypes(include='bool').columns if col != id_column]
if bool_columns:
    print(f'將 bool 欄位轉為 float32: {len(bool_columns)}')
    combined_data[bool_columns] = combined_data[bool_columns].astype('float32')

numeric_columns = [col for col in combined_data.select_dtypes(include='number').columns if col != id_column]
numeric_data = combined_data[numeric_columns].astype('float32')

argv_backup = sys.argv.copy()
sys.argv = [sys.argv[0], '--max_epochs', '10', '--batch_size', '256']  # or whatever args you need
try:
    remasker = ReMasker()
finally:
    sys.argv = argv_backup

ckpt_path = "checkpoints\\checkpoint_epoch_42.pth"  # adjust if needed
imputed_array = remasker.laod_param_fit_transform(numeric_data, ckpt_path)

imputed_numeric = pd.DataFrame(
    imputed_array,
    columns=numeric_columns,
    index=combined_data.index,
)
```

```python
train = pd.read_csv("home-credit-default-risk/application_train.csv")
test = pd.read_csv("home-credit-default-risk/application_test.csv")

train_count = len(train)
test_count = len(test)
total_count = train_count + test_count

if total_count != len(imputed_numeric):
    raise ValueError(f'填補後的資料列數 ({len(imputed_numeric)}) 與 train/test 總列數 ({total_count}) 不一致。')

train_ids = combined_data.iloc[:train_count][id_column].to_numpy()
test_ids = combined_data.iloc[train_count:][id_column].to_numpy()

imputed_train = imputed_numeric.iloc[:train_count].copy()
imputed_test = imputed_numeric.iloc[train_count:].copy()

imputed_train.insert(0, id_column, train_ids)
imputed_test.insert(0, id_column, test_ids)

print(f'填補後訓練資料列數: {imputed_train.shape[0]}, 欄位數: {imputed_train.shape[1]}')
print(f'填補後測試資料列數: {imputed_test.shape[0]}, 欄位數: {imputed_test.shape[1]}')
```

```python
target_df = pd.read_csv('home-credit-default-risk/application_train.csv', usecols=['SK_ID_CURR', 'TARGET'])
target_df['SK_ID_CURR'] = pd.to_numeric(target_df['SK_ID_CURR'], errors='coerce')
target_df = target_df.dropna(subset=['SK_ID_CURR', 'TARGET']).copy()
target_df['SK_ID_CURR'] = target_df['SK_ID_CURR'].astype('int64')

imputed_train_with_target = imputed_train.merge(target_df, on='SK_ID_CURR', how='left')
missing_targets = imputed_train_with_target['TARGET'].isna().sum()
if missing_targets:
    raise ValueError(f'有 {missing_targets} 筆訓練資料找不到 TARGET，請檢查 ID 對齊情況。')

output_path = 'imputed_train.csv'
imputed_train_with_target.to_csv(output_path, index=False)
print(f'已輸出含 TARGET 的訓練資料至 {output_path} ，共 {imputed_train_with_target.shape[0]} 列、{imputed_train_with_target.shape[1]} 欄。')
imputed_train_with_target.to_csv('imputed_train.csv', index=False)
```

```python
import pandas as pd
import numpy as np
from sklearn.feature_selection import mutual_info_regression
from sklearn.preprocessing import LabelEncoder

def compute_pmi(df):
    df = df.copy()

    # 將所有類別型特徵 label encode
    for col in df.columns:
        if df[col].dtype == 'object' or df[col].dtype.name == 'category':
            df[col] = LabelEncoder().fit_transform(df[col].astype(str))

    cols = df.columns
    n = len(cols)

    pmi_matrix = np.zeros((n, n))

    # 兩兩計算 mutual information
    for i in range(n):
        for j in range(n):
            if i == j:
                pmi_matrix[i, j] = 0
            else:
                # MI(X, Y) = MI(Y, X)，只算一邊
                mi = mutual_info_regression(
                    df[[cols[i]]], 
                    df[cols[j]], 
                    discrete_features='auto'
                )[0]
                pmi_matrix[i, j] = mi

    return pd.DataFrame(pmi_matrix, index=cols, columns=cols)

# ---- 使用示範 ----
df = pd.read_csv("home-credit-default-risk/application_train.csv")
pmi = compute_pmi(df)
print(pmi)
```
